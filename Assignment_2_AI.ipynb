```python
#Step 2: Import Libraries
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
#Step 3: Load GPT-2
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

model.eval()

print("GPT-2 loaded successfully!")
```

```python
#Step 4: Create the Original Prompt
prompt = "Artificial intelligence is changing the way we"

print("Original Prompt:", prompt)
```

```python
#Step 5: Create a Text Generation Function
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0, repetition_penalty=1.0):

    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        output = model.generate(
            input_ids=inputs["input_ids"],
            attention_mask=inputs["attention_mask"],
            max_new_tokens=50,
            do_sample=True,
            temperature=temperature,
            top_k=top_k,
            top_p=top_p,
            repetition_penalty=repetition_penalty,
            pad_token_id=tokenizer.eos_token_id
        )

    return tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
#Step 6: Experiment 1 — Temperature
output1 = generate_text(
    prompt,
    temperature=0.5
)

print(output1)
```

```python
#Step 7: Experiment 2 — Temperature
output2 = generate_text(prompt,
    temperature=1.0
)

print(output2)
```

```python
#Step 8: Experiment 3 — Top-k
output3 = generate_text(
    prompt,
    temperature=1.0,
    top_k=20
)

print(output3)
```

```python
#Step 9: Experiment 4 — Top-p
output4 = generate_text(
    prompt,
    temperature=1.0,
    top_p=0.7
)

print(output4)
```

```python
#Step 10: Experiment 5 — Repetition Penalty
output5 = generate_text(
    prompt,
    temperature=1.0,
    repetition_penalty=1.5
)

print(output5)
```

```python
#Step 11: Custom Prompt
custom_prompt = "Once upon a time, a young girl discovered a mysterious door and it was mysterious"

print("Custom Prompt:", custom_prompt)
```

```python
#Step 12: Generate 3 Different Versions
custom_output1 = generate_text(
    custom_prompt,
    temperature=0.5
)

print(custom_output1)
```

```python
custom_output2 = generate_text(
    custom_prompt,
    temperature=1.0
)

print(custom_output2)
```

```python
custom_output3 = generate_text(
    custom_prompt,
    temperature=1.5
)

print(custom_output3)
```

### Final Analysis

The experiments showed that generation parameters have a significant effect on GPT-2's output. Lower temperature generally produces more predictable and focused text, while higher temperature can produce more creative but sometimes less coherent text. Top-k and top-p control the number or probability range of words considered during generation. Repetition penalty helps reduce repeated words and phrases. Overall, changing these parameters allows us to control the balance between creativity, randomness, and coherence in generated text.
